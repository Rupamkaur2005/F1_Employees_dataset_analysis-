#using Groupby function

```python
type(df)
```

```python
df.groupby("Subscription Type")
```

Explanation:
. Pehla data nu Subscription Type de according 3 groups (Basic, Premium, Standard) vich divide kita janda hai.
. Fir har group di Monthly Revenue da total (Sum) calculate kita janda hai.

Output:
. Basic → Basic group de sare users di total Monthly Revenue.
. Premium → Premium group de sare users di total Monthly Revenue.
. Standard → Standard group de sare users di total Monthly Revenue.

```python
df.groupby("Subscription Type")["Monthly Revenue"].sum()
```

Explanation:
. Pehla data nu Subscription Type de according 3 groups (Basic, Premium, Standard) vich divide kita janda hai.
. Fir har group di Monthly Revenue di average (Mean) calculate kiti jandi hai.

Output:
. Basic → Basic group de users di average Monthly Revenue.
.Premium → Premium group de users di average Monthly Revenue.
.Standard → Standard group de users di average Monthly Revenue.

Conclusion:
Mean function har group di average Monthly Revenue dikhaunda hai.

```python
df.groupby("Subscription Type")["Monthly Revenue"].mean()
```
