```python
depar_count=df.groupby("Department").size().reset_index(name="Total Employees")
fig=px.bar(
    depar_count,
    x="Department",y="Total Employees",
    color="Department",
    text="Total Employees",
    title="Total Employees by Department",
    hover_data={
        "Department":True,
        "Total Employees":True
    }

    )
fig.update_traces(
textposition="outside",
marker_line_color="black",
marker_line_width=1.5


)

fig.update_layout(
    template="plotly_white",
    title_x=0.5,height=500,
    width=800,
    showlegend=True
    
)
fig.show()
```

Age Distribution of the Workforce

```python
fig=px.histogram(
    df,
    x="Age",
    nbins=10,
    title="Age Distribution of the Workforce",

    color_discrete_sequence=px.colors.qualitative.Pastel1,
    text_auto="True",
    hover_data={"Age":True}
)
fig.update_traces(
    marker_line_color="black",
    marker_line_width=1.9,


)
fig.update_layout(
    template="plotly_white",
    title_font=dict(size=22),
    height=500,
    width=800,
    title_x=0.5,
    showlegend=True,
    yaxis_title="Count of Employees",
    xaxis_title="Age Distribution of Employees",
    hoverlabel=dict(
        bgcolor="red",
        font_color="white"
    )

)
fig.show()
```

Gender Diversity Across Roles [Treemap]

```python
fig=px.treemap(
    df,
    path=["JobRole","Gender"],
    title="Gender Distribution Across Job Roles",
    color="Gender",
    color_discrete_sequence=["blue","pink"],
    hover_data={ 
    "JobRole":True,"Gender":True
    }
)
fig.update_traces(
    textfont=dict(size=15),
    marker_line_width=5,
    marker_line_color="black",
    root_color="lightgrey",textinfo="label+value"
    
) 
fig.update_layout(
    template="plotly_white",
    title_font=dict(size=22),
    height=700,
    width=1000,
    title_x=0.5,
    hoverlabel=dict(
        bgcolor="red",
        font_color="white"
    ))
fig.show()
```

Marital Status Breakdown[Pie Chart]  It helps HR understand how employees are divided into:
 Single, Married, and Divorced categories.

```python
marital_count=df.groupby("MaritalStatus").size().reset_index(name="Count")
fig=px.pie(
    marital_count,
    names="MaritalStatus",
    values="Count",
    title="Marital Status Breakdown",
    color="MaritalStatus",
    color_discrete_sequence=px.colors.qualitative.Dark24,
)
fig.update_traces(
    textinfo="label+percent",
    hole=0.4
)
fig.update_layout(
    title_font_size=40,
    height=700,
    width=900,
    template="plotly_dark"



)
```

[Employee Attrition OverView]

This shows the number of employees who left the company and who are still working.

```python
attrition_count=df.groupby("Attrition").size().reset_index(name="Count")
fig=px.bar(
    attrition_count,
    x="Attrition",
    y="Count",
    title="<b>Employee Attrition Overview</b>",
    color="Attrition",
    text_auto=True,
    )
fig.update_layout(
    title_x=0.5,
    xaxis_title="Attrition",
    yaxis_title="Number of Employees",
    template="plotly_white",
    height=500,
    width=700
)
fig.update_traces(
    textposition="outside"

)
fig.show()
```

[Employees Attrition by JobRole]

```python
employees_left=df[df["Attrition"]=="Yes"]
jobs=employees_left.groupby("JobRole").size().reset_index(name="Count")
fig=px.bar(
jobs,
x="JobRole",
y="Count",
title="Employees Left by Job Role",
text_auto=True,
color="JobRole",
color_discrete_sequence=px.colors.qualitative.Pastel
)
fig.update_traces(
textfont_size=12,
marker_line_color="black",
marker_line_width=3
)

fig.update_layout(
    title_x=0.4,
    xaxis_title="Job Role",
    yaxis_title="Number of Employees",
    hoverlabel=dict(
        bgcolor="white",
          font_color="red"  ),
          font_color="black",
          font_size=13

)
fig.show()
```

Attrition Trend by Years At Company

```python
employees_left=df[df["Attrition"]=="Yes"]
years=employees_left.groupby("YearsAtCompany").size().reset_index(name="count")
fig=px.line(
    years,
    x="YearsAtCompany",
    y="count",
    title="Attrition Trend Years by Company",
    markers=True
)
fig.update_traces(
    line_color="green",
    line_width=3,
    marker=dict(
        size=7,
        color="red"
    ),
    
)
fig.update_layout(
    title_x=0.4,
    xaxis_title="Years at Company",
    yaxis_title="Employees Left",
    
)


fig.show()
```
